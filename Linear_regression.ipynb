```python
import pandas as pd
df=pd.read_excel("/content/drive/MyDrive/Data Science/Data Set/tip-amount.xlsx")
df.head()

```

```python
# Meal column has no relevance on the output(Target) ,So drop it
df.drop('Meal',axis=1,inplace=True)
df.head()

```

```python
#to find the corelation between column
df.corr()
```

```python
# feature target splitting
y=df['Observed tip amount(yi)']
x=df.drop('Observed tip amount(yi)',axis=1)
```

```python
# train - test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
len(x_train),len(x_test),len(y_train),len(y_test)
```

```python
# creating linear regression model
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
model = lr.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
print(y_pred)
print(y_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
print(mean_squared_error(y_test, y_pred))
print(r2_score(y_test, y_pred))
```
